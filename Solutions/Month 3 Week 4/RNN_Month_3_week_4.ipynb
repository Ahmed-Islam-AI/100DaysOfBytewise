# **Recurrent Neural Networks**

1. **Question:** What are Recurrent Neural Networks, and how do they differ from traditional feedforward neural networks?
**Ans:** RNNs are a type of neural network designed to handle sequential data. Unlike traditional feedforward networks, RNNs use loops to allow information to persist, making them well-suited for tasks where context is crucial, like time series data or natural language processing.
### **Task**
**Working:** RNNs pass information through their layers over time. Each time step feeds the network not just the current input but also information from previous steps. This ability to retain memory is make it different from other networks.

2. **Stacking RNN Layers and Bi-directional Architecture**

* Stacking RNNs can allow the model to learn more complex patterns, as each layer processes the sequence at a higher level of abstraction.
* These networks run inputs in both forward and backward directions, allowing the model to capture information from both past and future time steps. This can significantly improve performance for tasks like text generation or speech recognition.
* While stacking and using bi-directional RNNs can enhance model performance, they can also increase training complexity and computational cost.

3. **Hybrid Architecture**

* A hybrid architecture in sequence modeling combines RNNs with other models, such as CNNs or Attention mechanisms. RNNs can capture temporal dependencies, while CNNs can help extract spatial features.
* **Example:** In a language translation task, an RNN could handle the sequential data, while an attention mechanism helps the model focus on specific parts of the input at different times, improving the models overall performance.

4. **Types of RNNs:**
* **Basic RNN:** Processes sequences one step at a time, passing hidden states between steps.
* **LSTM (Long Short-Term Memory):** Addresses the vanishing gradient problem in basic RNNs by introducing memory gates to manage what information to keep or discard.
* **GRU (Gated Recurrent Unit):** Similar to LSTM but with fewer parameters, making it faster while still handling long-range dependencies well.

# IMDB Movie Reviews Dataset

```python
#important libraries

import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.layers import Bidirectional
from tensorflow.keras.layers import Conv1D, MaxPooling1D
```

```python
# Loading the IMDB dataset
max_features = 10000  # Vocabulary size
max_len = 200  # Maximum review length
```

```python
# Loading the data and pad the sequences
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)
x_train = sequence.pad_sequences(x_train, maxlen=max_len)
x_test = sequence.pad_sequences(x_test, maxlen=max_len)
```

# Basic RNN Model

```python
# Building the RNN model
model_rnn = Sequential()
model_rnn.add(Embedding(max_features, 32, input_length=max_len))
model_rnn.add(SimpleRNN(32))
model_rnn.add(Dense(1, activation='sigmoid'))
```

```python
# Compile the model
model_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history_rnn = model_rnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.2)
```

```python
# Evaluate on test data
test_loss, test_acc = model_rnn.evaluate(x_test, y_test)
print(f"Test accuracy for Basic RNN: {test_acc}")
```

# Stacking RNN Layers & Bi-Directional RNNs

```python
# Building a stacked RNN model
model_stacked_rnn = Sequential()
model_stacked_rnn.add(Embedding(max_features, 32, input_length=max_len))
model_stacked_rnn.add(SimpleRNN(32, return_sequences=True))  # First RNN layer with output sequences
model_stacked_rnn.add(SimpleRNN(32))  # Second RNN layer
model_stacked_rnn.add(Dense(1, activation='sigmoid'))
```

```python
# Compile the model
model_stacked_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history_stacked_rnn = model_stacked_rnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.2)
```

```python
test_loss, test_acc = model_stacked_rnn.evaluate(x_test, y_test)
print(f"Test accuracy for Stacked RNN: {test_acc}")
```

### Bi-Directional RNN:

```python
# Building a bi-directional RNN model
model_bi_rnn = Sequential()
model_bi_rnn.add(Embedding(max_features, 32, input_length=max_len))
model_bi_rnn.add(Bidirectional(SimpleRNN(32)))  # Bi-directional RNN layer
model_bi_rnn.add(Dense(1, activation='sigmoid'))
```

```python
# Compile the model
model_bi_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
history_bi_rnn = model_bi_rnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.2)

# Evaluate the model on test data
test_loss, test_acc = model_bi_rnn.evaluate(x_test, y_test)
print(f"Test accuracy for Bi-Directional RNN: {test_acc}")
```

# Hybrid Architecture (CNN + RNN)

```python
# Build the Hybrid CNN + RNN model
model_hybrid = Sequential()
model_hybrid.add(Embedding(max_features, 32, input_length=max_len))
model_hybrid.add(Conv1D(64, 5, activation='relu'))
model_hybrid.add(MaxPooling1D(pool_size=4))
model_hybrid.add(SimpleRNN(32))
model_hybrid.add(Dense(1, activation='sigmoid'))
```

```python
# Compile the model
model_hybrid.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history_hybrid = model_hybrid.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.2)
```

```python
test_loss, test_acc = model_hybrid.evaluate(x_test, y_test)
print(f"Test accuracy for Hybrid CNN + RNN: {test_acc}")
```
